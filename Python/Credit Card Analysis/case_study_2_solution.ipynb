```python
cust_repay = pd.merge(left = cust_df,
              right = repay_df,
              left_on = 'Customer',
              right_on = 'Customer',
              how = 'inner')
```

```python
#1. In the above dataset,
#a. In case age is less than 18, replace it with mean of age values.
```

```python
mean_age = np.mean(cust_df['Age'])
def replace_age(x):
    if x < 18:
        x = mean_age
    return x
```

```python
cust_df['Age'] = cust_df['Age'].apply(replace_age)
cust_repay['Age'] = cust_repay['Age'].apply(replace_age)
cust_spend['Age'] = cust_spend['Age'].apply(replace_age)
```

```python
#b. In case spend amount is more than the limit, replace it with 50% of that customer’s limit.
#(customer’s limit provided in acquisition table is the per transaction limit on his card)
```

```python
cust_spend.head()
```

```python
cust_spend.loc[cust_spend['Amount'] > cust_spend['Limit'], 'Amount'] = 0.5 * cust_spend['Limit']
```

```python
#c. Incase the repayment amount is more than the limit, replace the repayment with the limit.
```

```python
cust_repay.head()
```

```python
cust_repay.loc[cust_repay['Amount'] > cust_repay['Limit'], 'Amount'] = cust_repay['Limit']
```

```python
#e. If the monthly rate of interest is 2.9%, what is the profit for the bank for each month?
#(Profit is defined as interest earned on Monthly Profit. Monthly Profit = Monthly repayment
#– Monthly spend. Interest is earned only on positive profits and not on negative amounts)
```

```python
trans = pd.merge(left = monthly_spend,
                 right = monthly_repay,
                 left_on = ['Customer', 'month'],
                 right_on = ['Customer', 'month'],
                 how = "inner")
```

```python
trans['Profit'] = trans['TotSpend'] - trans['TotalRepay']
trans
```

```python
trans['MonthlyBankProfit'] = trans['Profit'].apply(lambda x: x*0.29 if x>0 else 0)
trans
```

```python
#5. Write user defined PYTHON function to perform the following analysis:
#You need to find top 10 customers for each city in terms of their repayment amount by
#different products and by different time periods i.e. year or month. The user should be able
#to specify the product (Gold/Silver/Platinum) and time period (yearly or monthly) and the
#function should automatically take these inputs while identifying the top 10 customers.
```

```python
cust_spend.head()
```

```python
def find_top_ten():
    result = None
    product = input("Enter the name of the product: ")
    product = product.title()
    df = cust_spend.loc[cust_spend['Product'] == product]
    tp = input("Enter if you would like to sort the products (year/month): ")

    if tp == 'year':
        result = df.groupby(by = ['Customer', 'City', 'year']).agg(TotSpend = ('Amount', 'sum')).reset_index()
        result.sort_values(by = ['City','TotSpend'], ascending = [True,False])
    else:
        result = df.groupby(by = ['Customer', 'City', 'month']).agg(TotSpend = ('Amount', 'sum')).reset_index()
        result.sort_values(by = ['City','TotSpend'], ascending = [True,False])

    return result.head(10)
```
